```python
train_df = train_df.loc[:, ~train_df.columns.str.contains("^Unnamed")]
test_df = test_df.loc[:, ~test_df.columns.str.contains("^Unnamed")]
```

```python
X = train_df.drop("prognosis", axis=1)
y = train_df["prognosis"]

X_test = test_df.drop("prognosis", axis=1)
y_test = test_df["prognosis"]
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

accuracy = model.score(X_test, y_test)
print("Model Accuracy:", accuracy)
```

```python
import pandas as pd

def predict_disease(symptom_list):
    input_dict = {symptom: 0 for symptom in X.columns}
    
    for symptom in symptom_list:
        if symptom in input_dict:
            input_dict[symptom] = 1
    
    input_df = pd.DataFrame([input_dict])
    prediction = model.predict(input_df)
    return prediction[0]
```

```python
result = predict_disease(['itching', 'skin_rash', 'nodal_skin_eruptions'])
print("Predicted Disease:", result)
```

```python
import pandas as pd

def predict_disease_from_text(symptom_text):
    user_symptoms = [s.strip().lower().replace(" ", "_") 
                     for s in symptom_text.split(",")]
    

    input_dict = {symptom: 0 for symptom in X.columns}
    
    
    for symptom in user_symptoms:
        if symptom in input_dict:
            input_dict[symptom] = 1
    
    
    input_df = pd.DataFrame([input_dict])
    
    prediction = model.predict(input_df)
    return prediction[0]
```

```python
user_input = "weakness_in_limbs, fast_heart_rate, pain_during_bowel_movements, pain_in_anal_region"
result = predict_disease_from_text(user_input)

print("Predicted Disease:", result)
```
